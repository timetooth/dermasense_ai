```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv
```

```python
data = '/Users/study/Downloads/Ham10000/dataset/images'
meta = pd.read_csv('/Users/study/Downloads/Ham10000/dataset/metadata.csv')
```

```python
meta.head()
```

```python
root = './train'

for row in meta.iterrows():
    img = cv.imread(os.path.join(data, row[1]['image_id'] + '.jpg'))
    class_img = row[1]['dx']
    cv.imwrite(os.path.join(root, class_img + '_' + row[1]['image_id'] + '.jpg'), img)
```
